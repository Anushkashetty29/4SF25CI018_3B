```python
df = pd.read_csv("dataset.csv")

# Re-create 'new popularity' if you need it
df['new popularity'] = df['popularity'].astype(int)
```

```python
df.head()
```

```python
# Calculate average popularity for each genre
genre_insights = df.groupby('track_genre')['popularity'].mean().reset_index()

# Select only top 10 genres
genre_insights = genre_insights.nlargest(10, 'popularity')

# Plot bar graph
plt.figure(figsize=(10, 5))

sns.barplot(
    data=genre_insights,
    x='track_genre',
    y='popularity',
    palette='Blues'
)

plt.title('Top 10 Music Genres by Average Popularity')
plt.xlabel('Music Genre')
plt.ylabel('Average Popularity')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
popularity_insights = df.groupby('new popularity')['energy'].mean().reset_index()
plt.figure(figsize=(8, 4))

sns.lineplot(
    data=popularity_insights,
    x='new popularity',
    y='energy',
    marker='o',
    color='crimson'
)

plt.title('Average Energy by Popularity Level')
plt.xlabel('Popularity')
plt.ylabel('Average Energy')

plt.show()
```

```python
# Select only 10 popularity values
popularity_energy_insights = df.groupby('new popularity')['energy'].mean().reset_index().head(10)

plt.figure(figsize=(8, 4))

sns.barplot(
    data=popularity_energy_insights,
    x='new popularity',
    y='energy',
    hue='new popularity',
    palette='viridis',
    legend=False
)

plt.title('Average Track Energy by Popularity')
plt.xlabel('Popularity')
plt.ylabel('Average Energy')

plt.tight_layout()
plt.show()
```

```python
top_tracks = df.nlargest(10, 'popularity')

plt.figure(figsize=(12, 6))

sns.barplot(data=top_tracks, x='track_name', y='popularity')

plt.title('Top 10 Most Popular Spotify Tracks')
plt.xlabel('Track Name')
plt.ylabel('Popularity')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
genre_popularity = df.groupby('track_genre')['popularity'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))

sns.barplot(x=genre_popularity.index, y=genre_popularity.values)

plt.title('Top 10 Genres by Average Popularity')
plt.xlabel('Music Genre')
plt.ylabel('Average Popularity')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
# Select top 10 popularity values
popularity_energy_insights = df.groupby('new popularity')['energy'].mean().reset_index().head(10)

plt.figure(figsize=(8, 8))

plt.pie(
    popularity_energy_insights['energy'],
    labels=popularity_energy_insights['new popularity'],
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel')
)

plt.title('Track Energy Distribution by Popularity')
plt.legend(title='Popularity', loc='best')
plt.show()
```

```python
# Count explicit and non-explicit tracks
explicit_counts = df['explicit'].value_counts().sort_index()

# Plot pie chart
plt.figure(figsize=(7, 7))

plt.pie(
    explicit_counts,
    labels=['Non-Explicit', 'Explicit'],
    autopct='%1.1f%%',
    startangle=90,
    colors=['Green','Red'],
    explode=(0, 0.08),
    shadow=True
)

plt.title('Explicit vs Non-Explicit Spotify Tracks')

plt.show()
```
